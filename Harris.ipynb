```python
# Code for Semantic Segmentation with Thresholding
import cv2
import numpy as np

# Load image
img = cv2.imread('image.jpg')

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply binary thresholding (e.g., separate foreground and background)
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Show the result
cv2.imshow('Threshold Segmentation', thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Code for Instance Segmentation (Separate individual objects)

import cv2
import numpy as np

# Load the image
img = cv2.imread('image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Thresholding to get binary image
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Remove noise using morphological operations
kernel = np.ones((3, 3), np.uint8)
thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

# Marker for Watershed (basically labeling regions)
markers = cv2.connectedComponents(thresh)[1]
markers = markers + 1

# Perform Watershed segmentation
cv2.watershed(img, markers)

# Mark the boundaries
img[markers == -1] = [0, 0, 255]

cv2.imshow('Instance Segmentation', img)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Code for Instance Segmentation using Watershed
import cv2
import numpy as np

# Load the image
img = cv2.imread('image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Thresholding to get binary image
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Remove noise using morphological operations
kernel = np.ones((3, 3), np.uint8)
thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

# Marker for Watershed (basically labeling regions)
markers = cv2.connectedComponents(thresh)[1]
markers = markers + 1

# Perform Watershed segmentation
cv2.watershed(img, markers)

# Mark the boundaries
img[markers == -1] = [0, 0, 255]

cv2.imshow('Instance Segmentation', img)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Code for Basic Thresholding

import cv2

# Load image
img = cv2.imread('image.jpg')

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply binary thresholding
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Show the thresholded image
cv2.imshow('Thresholded Image', thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()


# Code for K-means Clustering Segmentation

import cv2
import numpy as np

# Load image
img = cv2.imread('image.jpg')

# Reshape the image into a 2D array of pixels
Z = img.reshape((-1, 3))

# Convert to float32 for K-means
Z = np.float32(Z)

# Criteria for K-means
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 100, 0.2)

# Apply K-means clustering
k = 3  # Number of clusters
_, labels, centers = cv2.kmeans(Z, k, None, criteria, 10, cv2.KMEANS_RANDOM_CENTERS)

# Convert back to 8 bit values
centers = np.uint8(centers)

# Map the labels to the centers
segmented_img = centers[labels.flatten()]

# Reshape back to the original image shape
segmented_img = segmented_img.reshape(img.shape)

cv2.imshow('K-means Segmentation', segmented_img)
cv2.waitKey(0)
cv2.destroyAllWindows()


# Code for Edge-based Segmentation using Canny
import cv2

# Load image
img = cv2.imread('image.jpg')

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Canny edge detection
edges = cv2.Canny(gray, 100, 200)

# Show the edge-detected image
cv2.imshow('Canny Edge Detection', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()


# Region-based Segmentation (Region Growing)
from skimage import measure
import numpy as np
import cv2

# Load image
img = cv2.imread('image.jpg', cv2.IMREAD_GRAYSCALE)

# Threshold the image to make it binary
_, thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

# Label connected components
labels = measure.label(thresh)

# Show the labeled regions
cv2.imshow('Region Segmentation', labels.astype(np.uint8) * 255)  # Scaling for display
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Harris
import cv2
import numpy as np

# Step 1: Load the image
img = cv2.imread('image.jpg')

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Step 2: Compute the gradients (Ix, Iy)
gray = np.float32(gray)  # Convert to float32 for precision
Ix = cv2.Sobel(gray, cv2.CV_32F, 1, 0, ksize=3)  # Gradient in X
Iy = cv2.Sobel(gray, cv2.CV_32F, 0, 1, ksize=3)  # Gradient in Y

# Step 3: Compute the structure matrix M
Ix2 = Ix**2
Iy2 = Iy**2
Ixy = Ix * Iy

# Apply Gaussian smoothing to the gradients (to reduce noise)
Ix2 = cv2.GaussianBlur(Ix2, (5, 5), 0)
Iy2 = cv2.GaussianBlur(Iy2, (5, 5), 0)
Ixy = cv2.GaussianBlur(Ixy, (5, 5), 0)

# Step 4: Compute the Harris response (R)
k = 0.04  # Sensitivity constant
detM = Ix2 * Iy2 - Ixy**2
traceM = Ix2 + Iy2
R = detM - k * (traceM**2)

# Step 5: Apply threshold to find strong corners
threshold = 0.01 * R.max()
corner_response = R > threshold

# Step 6: Mark the corners on the image
img[corner_response] = [0, 0, 255]  # Red color for corners

# Show the result
cv2.imshow('Harris Corner Detection', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
