# Image Augmentation:
As the amount of data increases the performance of the deep learning model also improves. But acquiring a massive amount of data comes with its own challenges. Every time it is not possible to have a large amount of data to feed the deep learning network.

The problem with the lack of a good amount of data is that the deep learning model might not learn the patterns or the functions from the data and hence it might not perform well.

So how can we deal with this? So instead of spending days manually collecting the data, we may use Image Augmentation techniques.

Image augmentation is a technique of altering the existing data to create some more data for the model training process. In other words, it is the process of artificially expanding the available dataset for training a deep learning model.

```python
import cv2
import numpy as np
```

# Identity Augmentation

```python
Original_Image = cv2.imread("Img.jpeg")
cv2.namedWindow("Original Image", cv2.WINDOW_AUTOSIZE)
cv2.imshow("Original Image", Original_Image)
cv2.waitKey(0)                   # Wait for a key press
cv2.destroyAllWindows()           # Close the window
```

The image above the is only the original image, and the rest of the images after this are generated from the original training image.

# Scaling

```python
# Set scaling factors
scale_x = 1.5  # Scale width
scale_y = 1.5  # Scale height

# Resize the image
scaled_image = cv2.resize(Original_Image, None, fx=scale_x, fy=scale_y, interpolation=cv2.INTER_LINEAR)
                                # The interpolation parameter in cv2.resize() determines how pixel values are computed when resizing an image.
                                #  It computes the pixel value as a weighted average of the four nearest neighboring pixels.
# Save and display the scaled image
cv2.imwrite("Scaled_Image.jpg", scaled_image)
cv2.namedWindow("Scaled Image", cv2.WINDOW_AUTOSIZE)
cv2.imshow("Scaled Image", scaled_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Mirroring

```python
# Flip horizontally (mirror effect)
mirrored_image = cv2.flip(Original_Image, 1)  # 1 flips along the vertical axis

# Save and display the mirrored image
cv2.imwrite("Mirrored_Image.jpg", mirrored_image)
cv2.namedWindow("Mirrored Image", cv2.WINDOW_AUTOSIZE)
cv2.imshow("Mirrored Image", mirrored_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Rotation

```python
# Get the image dimensions
(h, w) = Original_Image.shape[:2]  # (Height, width, channels) from that slice the first two elements height and width

# Define the center of rotation
center = (w // 2, h // 2)

# Set the rotation angle (e.g., 90 degrees)
angle = -90   # Positive angle is counter-clockwise
             # Negative angle is clockwise

# Compute the rotation matrix
rotation_matrix = cv2.getRotationMatrix2D(center, angle, 1.0)

# Compute the new bounding box dimensions after rotation
cos_val = abs(rotation_matrix[0, 0])
sin_val = abs(rotation_matrix[0, 1])

new_w = int((h * sin_val) + (w * cos_val))
new_h = int((h * cos_val) + (w * sin_val))

# Adjust the rotation matrix to account for translation
rotation_matrix[0, 2] += (new_w - w) // 2
rotation_matrix[1, 2] += (new_h - h) // 2

# Perform the rotation with the updated size
rotated_image = cv2.warpAffine(Original_Image, rotation_matrix, (new_w, new_h))

# Save and display the rotated image
cv2.imwrite("Rotated_Image.jpg", rotated_image)
cv2.imshow("Rotated Image", rotated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Translation

```python
# Define translation in x and y directions
tx, ty = 100, 50  # Move 100 pixels right and 50 pixels down

# Create the translation matrix
translation_matrix = np.float32([[1, 0, tx], [0, 1, ty]])

# Get image dimensions
(h, w) = Original_Image.shape[:2]

# Compute new image dimensions to fit the translated image
new_w, new_h = w + abs(tx), h + abs(ty)

# Create the translation matrix
translation_matrix = np.float32([[1, 0, tx], [0, 1, ty]])

# Apply the translation with the expanded canvas
translated_image = cv2.warpAffine(Original_Image, translation_matrix, (new_w, new_h))

# Save and display the translated image
cv2.imwrite("Translated_Image.jpg", translated_image)
cv2.imshow("Translated Image", translated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Shear (vertical)
Shearing is a transformation that skews the image in a particular direction. In a vertical shear, the image is slanted left or right along the x-axis, while the y-coordinates remain unchanged.

```python
# Get image dimensions
(h, w) = Original_Image.shape[:2]

# Define the shear factor (controls the amount of slant)
shear_factor = 0.3  # Positive values shear upwards, negative values shear downwards

# Create the shear matrix for vertical shear
shear_matrix = np.float32([[1, shear_factor, 0], [0, 1, 0]])

# Compute new image width and height (expanding to fit the sheared image)
new_w = int(w + abs(shear_factor) * h)  # Width increases due to shear
new_h = h  # Height remains the same

# Apply the shear transformation
sheared_image = cv2.warpAffine(Original_Image, shear_matrix, (new_w, new_h))

# Save and display the sheared image
cv2.imwrite("Sheared_Image_Vertical.jpg", sheared_image)
cv2.imshow("Sheared Image (Vertical)", sheared_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Shear (horizontal)
In a horizontal shear, the image is slanted up or down along the y-axis, while the x-coordinates remain unchanged.

```python
# Get image dimensions
(h, w) = Original_Image.shape[:2]

# Define the shear factor (controls the amount of slant)
shear_factor = 0.3  # Positive values shear right, negative values shear left

# Create the shear matrix for horizontal shear
shear_matrix = np.float32([[1, 0, 0], [shear_factor, 1, 0]])

# Compute new image width and height (expanding to fit the sheared image)
new_w = w  # Width remains the same
new_h = int(h + abs(shear_factor) * w)  # Height increases due to shear

# Apply the shear transformation
sheared_image = cv2.warpAffine(Original_Image, shear_matrix, (new_w, new_h))

# Save and display the sheared image
cv2.imwrite("Sheared_Image_Horizontal.jpg", sheared_image)
cv2.imshow("Sheared Image (Horizontal)", sheared_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
