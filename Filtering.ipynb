```python
import cv2
import numpy as np
```

```python
Original_Image = cv2.imread("Img.jpeg")
cv2.imshow("Original Image", Original_Image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Mean Filter (Box Filter)
# Define the kernel (3x3 Mean Filter)
kernel = np.ones((3, 3), np.float32) / 9

# Apply the filter
mean_filtered = cv2.filter2D(Original_Image, -1, kernel)

# Save and display the scaled image
cv2.imwrite("Mean_Filtered_Image.jpg", mean_filtered)
cv2.imshow("Mean Filtered Image", mean_filtered)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Apply Gaussian filter (3x3 kernel, standard deviation auto-determined)
gaussian_filtered = cv2.GaussianBlur(Original_Image, (3, 3), 0)
                                                      # size of the kerne
                                                             # standard deviation (0 = auto-determined)
# sigmaX = 0 → OpenCV calculates a suitable value based on the 5×5 kernel size.
# sigmaY is not given, so it is set equal to sigmaX.
# Result: A smooth blur with an automatically optimized standard deviation.

# Save and display the filtered image
cv2.imwrite("Gaussian_Filtered_Image.jpg", gaussian_filtered)
cv2.imshow("Gaussian Filtered Image", gaussian_filtered)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Apply the Median Filter with a 3x3 kernel
median_filtered = cv2.medianBlur(Original_Image, 3)

# Save and display the filtered image
cv2.imwrite("Median_Filtered_Image.jpg", median_filtered)
cv2.imshow("Median Filtered Image", median_filtered)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Read the original image in grayscale
Original_Image = cv2.imread("Img.jpeg", cv2.IMREAD_GRAYSCALE)
# Save and display the grayscale image
cv2.imwrite("Grayscale_Image.jpg", Original_Image)
cv2.imshow("Grayscale Image", Original_Image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Apply Sobel filter in X direction (detects vertical edges)
sobel_x = cv2.Sobel(Original_Image, cv2.CV_64F, 1, 0, ksize=3)
         #cv2.Sobel(src, ddepth, dx, dy, ksize)
                         #Specifies the depth (data type) of the output image.
                         #cv2.CV_64F means 64-bit floating point (float64).
                        #This prevents loss of negative values when computing gradients.
                                  #dx = 1, dy = 0 → Applies Sobel filter in the X direction (detects vertical edges).
# Apply Sobel filter in Y direction (detects horizontal edges)
sobel_y = cv2.Sobel(Original_Image, cv2.CV_64F, 0, 1, ksize=3)

# Convert back to 8-bit image
sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

# Combine both gradients
sobel_combined = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

# Save and display the results
cv2.imwrite("Sobel_X.jpg", sobel_x)
cv2.imwrite("Sobel_Y.jpg", sobel_y)
cv2.imwrite("Sobel_Combined.jpg", sobel_combined)

cv2.imshow("Sobel X", sobel_x)
cv2.imshow("Sobel Y", sobel_y)
cv2.imshow("Sobel Combined", sobel_combined)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Apply Laplacian filter (Second-order derivative)
laplacian_filtered = cv2.Laplacian(Original_Image, cv2.CV_64F, ksize=3)

# Convert back to 8-bit format (for proper display)
laplacian_filtered = cv2.convertScaleAbs(laplacian_filtered)
                     #Converts the floating-point image to 8-bit (uint8) for proper display.
# Save and display the filtered image
cv2.imwrite("Laplacian_Filtered_Image.jpg", laplacian_filtered)
cv2.imshow("Laplacian Filtered Image", laplacian_filtered)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Define Prewitt kernels manually
prewitt_x = np.array([[-1, 0, 1], 
                       [-1, 0, 1], 
                       [-1, 0, 1]], dtype=np.float32)

prewitt_y = np.array([[-1, -1, -1], 
                       [0,  0,  0], 
                       [1,  1,  1]], dtype=np.float32)

# Apply Prewitt filter in X direction
prewitt_x_filtered = cv2.filter2D(Original_Image, -1, prewitt_x)
                                                  #-1 → Same depth as input image (keeps the same data type as Original_Image).
                                                      #prewitt_x detects horizontal edges.
# Apply Prewitt filter in Y direction
prewitt_y_filtered = cv2.filter2D(Original_Image, -1, prewitt_y)

# Combine both X and Y edges using absolute sum
prewitt_combined = cv2.addWeighted(prewitt_x_filtered, 0.5, prewitt_y_filtered, 0.5, 0) # Combines both X and Y edge-detected images using weighted sum.
                  #cv2.addWeighted(src1, alpha, src2, beta, gamma)
                                         #alpha = 0.5 → Weight of prewitt_x_filtered (50% contribution).

# Save and display the results
cv2.imwrite("Prewitt_X.jpg", prewitt_x_filtered)
cv2.imwrite("Prewitt_Y.jpg", prewitt_y_filtered)
cv2.imwrite("Prewitt_Combined.jpg", prewitt_combined)

cv2.imshow("Prewitt X", prewitt_x_filtered)
cv2.imshow("Prewitt Y", prewitt_y_filtered)
cv2.imshow("Prewitt Combined", prewitt_combined)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
